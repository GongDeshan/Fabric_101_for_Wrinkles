```python
import sympy as sp
from sympy import symbols
from sympy import Heaviside
from SymEigen import *

def step(x):
    return Heaviside(x, 1)

Gen = EigenFunctionGenerator()
Gen.MacroBeforeFunction('__host__ __device__')
Gen.DisableLatexComment()

Xn = Eigen.Vector('Xn', 6)
Cl = Gen.Closure(Xn)

P0 = Xn[0:3,:]
P1 = Xn[3:6,:]

ex = sp.Matrix([1,0,0])
ey = sp.Matrix([0,1,0])

n = P1 - P0
cross_x = n.cross(ex)
cross_y = n.cross(ey)
s = cross_x.dot(cross_x) - cross_y.dot(cross_y)
s

def Normalized(v):
    return v / sqrt(v.dot(v))
```

```python

e1 = step(s) * Normalized(cross_x) + (1 - step(s)) * Normalized(cross_y)
# e1
```

```python

e2 = Normalized(n.cross(e1))
# e2
```

```python
var_e1 = Eigen.Vector('e1', 3)
var_e2 = Eigen.Vector('e2', 3)

Basis = sp.Matrix([var_e1.T, var_e2.T])
Basis
```

```python
ClosestMatrix = sp.Matrix.zeros(3, 6)
ClosestMatrix[0:3, 0:3] = sp.Matrix.eye(3)
ClosestMatrix[0:3, 3:6] = -sp.Matrix.eye(3)

T = Basis @ ClosestMatrix

VecT = Sym.Vectorize(T)
VecT
```

```python
VecT = Subs(VecT, [(var_e1, e1), (var_e2, e2)])
```

```python

dVecTdXn = VecDiff(VecT, Xn)
```

```python
dTdXn = Sym.Vectorize(dVecTdXn)
```

```python

s = Cl('dTdXn', dTdXn)
print(s)
```
